script to sanity check reshaping processes within models

# UNet

```python
unet = models.ClimSimUNet()

x = torch.arange(0, 45*2+4).unsqueeze(0)

print(x.shape)
print(x)
```

```python
y = unet._reshape_from_standard_format(x)

print(y.shape)
print(y[0, 0, :])
print(y[0, 1, :])
print(y[0, 2, :])
print(y[0, 3, :])

```

```python
r = torch.arange(0, 45*2+8).unsqueeze(0)

y1 = r[:, 0:45]
y2 = r[:, 45:90]
y3 = torch.ones(1, 45) * 90
y4 = torch.ones(1, 45) * 91
y5 = torch.ones(1, 45) * 92
y6 = torch.ones(1, 45) * 93
y7 = torch.ones(1, 45) * 94
y8 = torch.ones(1, 45) * 95
y9 = torch.ones(1, 45) * 96
y10 = torch.ones(1, 45) * 97

y = torch.stack([y1, y2, y3, y4, y5, y6, y7, y8, y9, y10]).transpose(1, 0)
print(y.shape)
# print(y)
```

```python
out = unet._reshape_to_standard_format(y)

print(out.shape)
print(out)
```

# Squeezeformer

```python

squeezeformer = models.SqueezeFormer(in_dim=6, embed_dim=64, head_dim=256, out_dim=10)

x = torch.arange(0, 45*2+4).unsqueeze(0)

print(x.shape)
print(x)
```

```python
y = squeezeformer._reshape_from_standard_format(x)

print(y.shape)
print(y[0, :, 0])
print(y[0, :, 1])
print(y[0, :, 2])
print(y[0, :, 3])
```

```python
r = torch.arange(0, 45*2+8).unsqueeze(0)

y1 = r[:, 0:45]
y2 = r[:, 45:90]
y3 = torch.ones(1, 45) * 90
y4 = torch.ones(1, 45) * 91
y5 = torch.ones(1, 45) * 92
y6 = torch.ones(1, 45) * 93
y7 = torch.ones(1, 45) * 94
y8 = torch.ones(1, 45) * 95
y9 = torch.ones(1, 45) * 96
y10 = torch.ones(1, 45) * 97

y = torch.stack([y1, y2, y3, y4, y5, y6, y7, y8, y9, y10]).permute(1, 2, 0)
print(y.shape)
# print(y)
```

```python
out = squeezeformer._reshape_to_standard_format(y)

print(out.shape)
print(out)
```
